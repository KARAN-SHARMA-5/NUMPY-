# joining numpy arrays

```python
arr1 = np.array([1,2,3])
arr2=  np.array([4,5,6])
arr=np.concatenate ((arr1,arr2))
print(arr)
```

```python
arr1=np.array([[1,2,3,4],[5,6,7,8]])
arr2=np.array([[1,2,3,4],[5,6,7,8]])
arr=np.concatenate((arr1,arr2),axis=0)
print(arr)
```

```python
arr1=np.array([[1,2,3,4],[5,6,7,8]])
arr2=np.array([[1,2,3,4],[5,6,7,8]])
arr=np.concatenate((arr1,arr2),axis=1)
print(arr)
```

# ravel & flaten

```python
import numpy as np
m=np.array([[1,2,3],[5,6,7]])
n=m.ravel()
print(n)
```

```python
c=np.array([[[[1,2,3,4],[2,3,4,5],[5,6,7,8]]]])
d=c.ravel()
print(d)
print(d.ndim)
```

# unique function

```python
k=np.array([12,1,4,6,7,9,3,4,5,22,33,7,6,9,1,2,12])
z=np.unique(k)
print(z)
```

```python
k=np.array([12,1,4,6,7,9,3,4,5,22,33,7,6,9,1,2,12])
z=np.unique(k,return_index=True)
print(z)
```
